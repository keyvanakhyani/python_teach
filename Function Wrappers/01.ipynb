```python
def greet(name):
    return f"Hello, {name}!"

# Assign function to variable
say_hello = greet
print(say_hello("Alice"))
```

```python
def outer_function():
    message = "Hello"
    
    def inner_function():
        print(message)
    
    inner_function()

outer_function()
```

2.3 Closures

```python
# way 1 Returns the function object itself

def multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply

times_three = multiplier(3)  # times_three IS a function
times_three(10)              # Now we call it → Output: 30
```

```python
# way 1 Executes/calls the function immediately

def multiplier(factor):
    def multiply():
        return 10 * factor
    return multiply()

times_three = multiplier(3)  # times_three IS a function
times_three
```

3. Creating Simple Wrappers

# Basic Wrapper Function
my_wrapper takes a function as an argument
Inside my_wrapper, we define inner that wraps the original function
inner adds behavior before and after calling func()
my_wrapper returns the inner function
We call the wrapped version to see the added behavior

```python
def my_wrapper(func):
    def inner():
        print("Before function execution")
        func()
        print("After function execution")
    return inner

def say_hello():
    print("Hello!")

# Wrap the function
wrapped_hello = my_wrapper(say_hello)
print(wrapped_hello)
wrapped_hello()
```

4. Understanding Decorators

4.1 What are Decorators?
Decorators are a syntactic sugar for applying wrappers. Instead of manually wrapping functions, we use the @ symbol.

```python
def my_decorator(func):
    def wrapper():
        print("Something before")
        func()
        print("Something after")
    return wrapper

@my_decorator
def say_goodbye():
    print("Goodbye!")

say_goodbye()
```

```python
# Using decorator syntax
@my_decorator
def function():
    print('my_decorator')
    pass

# Manual wrapping
def function():
    print("Manual wrapping")
    pass

function1 = my_decorator(function)
function1()
print(8*"*")
function()
```

5. Preserving Function Metadata

5.1 The Problem
When we wrap functions, we lose their metadata:

```python
def decorator(func):
    """This is the docstring for decorator"""
    def wrapper():
        return func()
    return wrapper

@decorator
def original():
    """This is the docstring for original"""
    pass

print(original.__name__)
print(original.__doc__)
print(decorator.__name__)
print(decorator.__doc__)
```

5.2 Solution: functools.wraps
Use functools.wraps to preserve metadata:

```python
from functools import wraps

def decorator(func):
    """This is the docstring for decorator"""
    @wraps(func)
    def wrapper():
        return func()
    return wrapper

@decorator
def original():
    """This is the docstring"""
    pass

print(original.__name__)
print(original.__doc__)
print(decorator.__name__)
print(decorator.__doc__)
```

6. Wrappers with Arguments

6.1 Handling Function Arguments
Use *args and **kwargs to handle any number of arguments:

```python
from functools import wraps

def log_arguments(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Arguments: {args}")
        print(f"Keyword arguments: {kwargs}")
        result = func(*args, **kwargs)
        print(f"Result: {result}")
        return result
    return wrapper

@log_arguments
def add(a, b):
    return a + b

add(5, 3)
```
